```python
import pandas as pd

# Read the CSV file
df = pd.read_csv("class1_student_habits.csv")

# Select the numerical columns we want to examine
columns = [
    "study_hours_per_week",
    "sleep_hours_per_night",
    "attendance_rate",
    "practice_problems_per_week",
    "commute_minutes",
    "quiz_average",
    "exam_score"
]

# Calculate descriptive statistics
stats = df[columns].describe()

# Add the median, since describe() does not label it explicitly as "median"
stats.loc["median"] = df[columns].median()

# Keep only the requested statistics
stats = stats.loc[
    ["count", "mean", "median", "std", "min", "max"]
]

print(stats)
```

2. The max amount of hours spent studying per week for any one student was 17.8
3. The max amount of hours spent sleeping by any one student was 10 hours

```python
import pandas as pd

df = pd.read_csv("class1_student_habits.csv")
```

```python
filtered_students = df[
    (df["study_hours_per_week"] >= 10) &
    (df["attendance_rate"] >= 90)
]

print(
    filtered_students[
        [
            "study_hours_per_week",
            "attendance_rate",
            "quiz_average",
            "exam_score"
        ]
    ]
)
```

```python
top_10_exam_scores = df.sort_values(
    by="exam_score",
    ascending=False
).head(10)

print(
    top_10_exam_scores[
        [
            "exam_score",
            "quiz_average",
            "study_hours_per_week",
            "attendance_rate",
            "sleep_hours_per_night"
        ]
    ]
)
```

```python
study_group_stats = df.groupby("study_group")["exam_score"].agg(
    student_count="count",
    mean_exam_score="mean",
    median_exam_score="median"
)

print(study_group_stats)
```

Study per hours and attendance rate overall did not reliably determine how well a student was going to perform on a exam. Some of these columns really don't describe how good the quality of their study time and how much time a student actually understood the content when attending, ect.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("class1_student_habits.csv")

plt.figure(figsize=(8, 5))

plt.hist(
    df["exam_score"].dropna(),
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

It looks like overall from class 1, most students scored above 60 for their final exam grade. From this, it seems like the median is between 70-80.

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["study_hours_per_week"],
    df["exam_score"],
    alpha=0.7
)

plt.title("Study Hours per Week vs. Exam Score")
plt.xlabel("Study Hours per Week")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()
```

From this scatter plot, it follows that people who dedicated more time studying also happened to score higher for their exam  scores, suggesting a positive correlation. Meaning those who spent more time studying did better overall on the exam.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(
    df["quiz_average"],
    df["exam_score"],
    alpha=0.7
)

plt.title("Quiz Average vs. Exam Score")
plt.xlabel("Quiz Average")
plt.ylabel("Exam Score")

plt.tight_layout()
plt.show()
```

This scatter follows a linear trend of scoring a similar score from your quiz average to your exam score. So people who were already doing decently well in class also happened to score higher on the exam.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("class1_student_habits.csv")

df.plot.scatter(
    x="practice_problems_per_week",
    y="exam_score"
)

plt.title("Practice Problems vs Exam Score")
plt.xlabel("Practice Problems per Week")
plt.ylabel("Exam Score")
plt.show()
```

There seems to be a linear relationship between practice_problems and exam_score. There is a slight linear pattern suggesting a positive correlation between practice problems and exam scores. Meaning those who studied tended to score higher, this graph is displayed this way because of the one outlier who studied for almost 80 hours.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("class1_student_habits.csv")

df.plot.scatter(
    x="sleep_hours_per_night",
    y="study_hours_per_week"
)

plt.title("Sleep Hours vs Study Hours")
plt.xlabel("Sleep Hours per Night")
plt.ylabel("Study Hours per Week")
plt.show()
```

It doesn't seem like there is much of a linear correlation between Study Hours Per Week to Sleep hours per night. From this, I can promptly conclude that how many hours a student sleep has no relation to how many hours they may spend studying that week.
